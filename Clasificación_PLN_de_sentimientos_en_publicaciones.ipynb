## **Importamos librerias y leemos el .csv**

```python
import pandas as pd
df = pd.read_csv('dataset_train.csv')
df.head()
```

## **Verificaciones de nulo**

```python
df.info()
df.isnull().sum()
```

## **Verificamos si hay datos vacios**

```python
print("Cantidad de filas antes de eliminar duplicado", df.shape)
df.drop_duplicates(inplace=True)
print("Cantidad de filas después de eliminar duplicado", df.shape)
```

## **Entrenamos el modelo**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df['text'], df['sentimento'], test_size=0.10)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizador = CountVectorizer()
X_train_vec = vectorizador.fit_transform(X_train)
X_test_vec = vectorizador.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

modelo = LogisticRegression()
modelo.fit(X_train_vec, y_train)

predit = modelo.predict(X_test_vec)

accuracy = accuracy_score(y_test, predit)
print("Porcentaje de acierto:", accuracy * 100)
```

## **Leemos el .csv sin la etiqueta objetivo**

```python
pruebas = pd.read_csv('dataset_test_sin_sentimento.csv')
pruebas.head()
```

## **Generamos el archivo .csv**

```python
X_test_vec = vectorizador.transform(pruebas['text'])

predict = modelo.predict(X_test_vec)

output = pd.DataFrame({'ID': pruebas['id'], 'etiqueta': predict})
output.to_csv('submission.csv', index=False)

print("Archivo creado con éxito")
```
